```python
def parse_row(row):
    row['age'] = int(row['age'])
    row['fee_inr'] = int(row['fee_inr'])
    row['prior_no_shows'] = int(row['prior_no_shows'])
    row['distance_km'] = float(row['distance_km'])
    return row

accepted = []
rejected = []

# Corrected filename from "Appintment.csv" to "Appointments capstone.csv"
with open("Appointments capstone.csv", "r", newline="") as file:
    reader = csv.DictReader(file)

    for row in reader:
        try:
            parsed_row = parse_row(dict(row))
            accepted.append(parsed_row)
        except ValueError:
            rejected.append(row)

with open("rejected_rows.json", "w") as json_file:
    json.dump(rejected, json_file, indent=2)

print("Total accepted:", len(accepted))
print("Total rejected:", len(rejected))
```

```python
import requests
import pandas as pd

URL = 'https://api.open-meteo.com/v1/forecast'

temperatures = []

for index, row in clinics.iterrows():

    params = {
        "latitude": row['latitude'],
        "longitude": row['longitude'],
        "hourly": "temperature_2m",
        "forecast_days": 1
    }

    temp_c = None

    for attempt in range(2):

        try:
            response = requests.get(URL, params=params, timeout=10)

            if response.status_code == 200:

                data = response.json()

                if ('hourly' in data and
                    'temperature_2m' in data['hourly'] and
                    len(data['hourly']['temperature_2m']) > 0):

                    temp_c = data['hourly']['temperature_2m'][0]
                    break

                else:
                    print(row['clinic_name'], "Missing hourly temperature")
                    break

            else:
                print(row['clinic_name'], "API status:", response.status_code)
                break

        except requests.exceptions.RequestException as e:

            if attempt == 0:
                print(row['clinic_name'], "Request failed. Retrying...")
            else:
                print(row['clinic_name'], "Failed after retry:", e)

    temperatures.append(temp_c)

clinics['temp_c'] = temperatures

clinics.to_csv("clinics_enriched.csv", index=False)
```
